# Weather Data Analysis
Query PostgreSQL → Analyse → Visualise

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sqlalchemy import create_engine

engine = create_engine('postgresql://deuser:depass123@localhost:5432/de_lab')
df = pd.read_sql('SELECT * FROM weather_raw ORDER BY ingested_at DESC LIMIT 500', engine)
print(f'Records loaded: {len(df)}')
df.head()
```

```python
# Temperature comparison across cities
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Bar chart — avg temperature
city_avg = df.groupby('city')['temperature_c'].mean().sort_values(ascending=False)
axes[0].bar(city_avg.index, city_avg.values, color='steelblue')
axes[0].set_title('Average Temperature by City')
axes[0].set_ylabel('Temperature (°C)')
axes[0].tick_params(axis='x', rotation=45)

# Box plot — temperature distribution
df.boxplot(column='temperature_c', by='city', ax=axes[1])
axes[1].set_title('Temperature Distribution')
axes[1].set_xlabel('City')

plt.tight_layout()
plt.savefig('weather_analysis.png', dpi=150)
plt.show()
print('Chart saved!')
```

```python
# Correlation heatmap
numeric_cols = ['temperature_c','humidity_pct','pressure_hpa','wind_speed_ms','cloudiness_pct']
corr = df[numeric_cols].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', center=0)
plt.title('Weather Metrics Correlation')
plt.tight_layout()
plt.show()
```
